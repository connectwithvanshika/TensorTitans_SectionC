```python
from sklearn.metrics import (
    average_precision_score, precision_score,
    recall_score, f1_score, roc_auc_score
)

# DEMO ONLY: synthetic labels and probabilities.
y_true = [0, 0, 0, 1, 1, 1]
y_prob = [0.1, 0.2, 0.4, 0.8, 0.6, 0.9]
threshold = 0.5
y_pred = [int(p >= threshold) for p in y_prob]

print("PR-AUC / Average Precision:",
      average_precision_score(y_true, y_prob))
print("Precision:", precision_score(y_true, y_pred))
print("Recall:", recall_score(y_true, y_pred))
print("F1:", f1_score(y_true, y_pred))
print("ROC-AUC:", roc_auc_score(y_true, y_prob))

# TODO: Replace demo labels/probabilities with real predictions.
# TODO: Report test metrics only after finalizing model selection.
```
