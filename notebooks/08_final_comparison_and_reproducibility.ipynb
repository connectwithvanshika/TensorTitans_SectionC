## 1. Setup and Loading Final Benchmark Artifacts


```python
import sys
import os

repo_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
if repo_root not in sys.path:
    sys.path.append(repo_root)
if os.getcwd() not in sys.path:
    sys.path.append(os.getcwd())

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

results_dir = "results" if os.path.exists("results") else "../results"
metrics_file = os.path.join(results_dir, "model_comparison.csv")
manifest_file = "data/processed/split_manifest.csv" if os.path.exists("data/processed/split_manifest.csv") else "../data/processed/split_manifest.csv"

comparison_df = pd.read_csv(metrics_file)
manifest_df = pd.read_csv(manifest_file)
```

## 2. Consolidated Empirical Model Comparison Table
We present the final evaluation metrics across baselines, linear models, and tree-based gradient boosting architectures.


```python
display(comparison_df)

row_counts = pd.read_csv(os.path.join(results_dir, "metrics", "row_counts.csv"))
val_row = row_counts[row_counts["Split"] == "eval"]
test_row = row_counts[row_counts["Split"] == "test"]
val_fire_rate = val_row["Fire rows"].values[0] / val_row["Rows used"].values[0]
test_fire_rate = test_row["Fire rows"].values[0] / test_row["Rows used"].values[0]

plt.figure(figsize=(10, 5))
sns.barplot(x="Model", y="PR-AUC", hue="Split", data=comparison_df, palette="crest")
plt.axhline(val_fire_rate, color="gray", linestyle="--", label=f"Random guess, validation ({val_fire_rate:.4f})")
plt.axhline(test_fire_rate, color="black", linestyle=":", label=f"Random guess, test ({test_fire_rate:.4f})")
plt.title("Empirical PR-AUC Comparison Across Models and Partitions", fontsize=14, fontweight="bold")
plt.ylabel("PR-AUC (Average Precision)")
plt.legend()
plt.grid(True, axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(results_dir, "figures", "final_prauc_comparison.png"), dpi=200)
plt.show()
```
