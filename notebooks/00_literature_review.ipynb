## 1. Project Objective

Wildfires pose a growing threat to ecosystems, infrastructure, and human life, and their frequency and intensity are increasing due to climate change. The objective of this project is to predict **next-day wildfire spread** — specifically, whether a pixel or region adjacent to an active fire perimeter will burn within the following 24 hours.

To achieve this, we use a combination of **environmental input variables** drawn from publicly available remote-sensing and meteorological datasets. These variables include weather conditions (wind speed, temperature, humidity), topographic features (elevation, slope), vegetation type, drought indices, and historical fire masks. Two complementary modelling paradigms are explored:

- **Tabular ML models** (e.g., logistic regression, random forest, gradient boosting) applied to flattened pixel-level features.
- **Spatial deep learning models** (e.g., convolutional neural networks) that exploit the spatial structure of the input data.

The task is formulated as binary classification — *fire spread* vs. *no fire spread* — and performance is assessed using metrics that are robust to the strong class imbalance inherent in wildfire data.


## 2. Research Questions

This literature review is guided by the following three research questions:

1. **Which environmental factors influence wildfire spread?**  
   We aim to identify the most relevant predictors (weather, vegetation, topography, etc.) reported in existing literature.

2. **How effectively can ML models predict next-day wildfire spread?**  
   We examine the types of models evaluated, datasets used, and the reported predictive performance.

3. **Which evaluation metrics are appropriate for imbalanced wildfire data?**  
   Because fire-spread pixels are rare relative to non-fire pixels, standard accuracy is misleading. We look at what metrics the literature recommends.


## 3. Literature Review

The table below summarises three peer-reviewed papers directly relevant to this project. Papers were selected based on relevance to wildfire spread prediction, ML methodology, and use of environmental variables.

| # | Title & Year | Model / Approach | Key Findings | Limitation |
|---|---|---|---|---|
| 1 | *Next Day Wildfire Spread: A Machine Learning Dataset to Predict Wildfire Spreading From Remote-Sensing Data* (Huot et al., 2022) | Deep neural networks benchmarked against Logistic Regression and Random Forest on spatial remote-sensing data | A 12-variable environmental dataset (weather, topography, vegetation, drought, VIIRS fire mask) effectively frames next-day wildfire spread as a binary classification problem; deep CNNs improved upon tabular baselines. | Dataset is limited to the contiguous United States; deep models require significant compute and may not generalise to regions with different fire regimes. |
| 2 | *A Machine Learning-Based Approach for Wildfire Susceptibility Mapping* (Tonini et al., 2020) | Random Forest trained on geo-environmental variables (land cover, topography, road proximity) | Random Forest produced accurate susceptibility maps for a 21-year fire record in Liguria, Italy; land cover type and slope were the most important predictors. | Focuses on historical *susceptibility* rather than dynamic next-day *spread*; real-time weather conditions are not included as predictors. |
| 3 | *A Review of Machine Learning Applications in Wildfire Science and Management* (Jain et al., 2020) | Systematic review of ~300 ML studies across six wildfire problem domains | Random Forest, MaxEnt, and ANNs dominate the wildfire ML literature; deep learning is underutilised; class imbalance and data availability are recurring challenges. | Does not present original experimental results; many cited studies predate modern large-scale remote-sensing datasets and deep learning frameworks. |

## 4. Research Gap

Based on the three papers reviewed, a **preliminary research direction** can be identified. The limited number of papers is acknowledged as a constraint on establishing a definitive research gap.

Huot et al. (2022) introduce a benchmark for next-day wildfire spread prediction but evaluate only a small set of models (logistic regression, random forest, and one CNN architecture). Tonini et al. (2020) show that geo-environmental features are highly informative for spatial fire modelling, but their approach is static — it does not model the dynamic, day-by-day propagation of an active fire. Jain et al. (2020) observe that deep learning remains underutilised across the wildfire domain and that class imbalance is an unresolved methodological challenge.

Together, these papers suggest that **a systematic comparison of tabular ML models and spatial deep learning models, evaluated with imbalance-aware metrics on a standardised next-day fire spread dataset, has not yet been comprehensively documented**. The literature also lacks a clear analysis of how metric choice (AUC-ROC, F1, Area Under the Precision-Recall Curve) affects model selection under realistic class imbalance. This project is positioned as a preliminary investigation of these questions, using the Huot et al. (2022) dataset as a reproducible starting point.
